```python
# inicializar algumas bibliotecas e outras ferramentas de Python

from IPython.display import Audio as IAudio
from PIL import Image as PImage

from diffusers import ControlNetModel
from diffusers import DiffusionPipeline
from diffusers import StableDiffusionControlNetPipeline as ControlNetPipeline
from torch import cuda, float16 as tf16
from transformers import pipeline
from utils import show_object_predictions

DEVICE = "cuda" if cuda.is_available() else "cpu"
```

## Detecção de Objetos

```python
DETR_MODEL = "facebook/detr-resnet-50"
OWL_MODEL = "google/owlv2-base-patch16"

detr = pipeline(task="object-detection",
                model=DETR_MODEL,
                device=DEVICE)

owl = pipeline(task="zero-shot-object-detection",
               model=OWL_MODEL,
               device=DEVICE)
```

```python
img = PImage.open("mesa.jpg")
img.thumbnail((500, 500))

res = detr(img)
show_object_predictions(img, res)
```

```python
img = PImage.open("mesa.jpg")
img.thumbnail((500, 500))

res = owl(img, ["backpack", "bike"], threshold=0.12)
show_object_predictions(img, res)
```

## Processamento de Imagem

```python
DPT_MODEL = "Intel/dpt-large"

dpt = pipeline(task="depth-estimation",
               model=DPT_MODEL,
               device=DEVICE)
```

```python
img = PImage.open("mesa.jpg")
img.thumbnail((500, 500))

res = dpt(img)
display(res["depth"])
```

## Transcrição de Áudio

```python
WHISPER_MODEL = "openai/whisper-small"

whisper = pipeline(task="automatic-speech-recognition",
                   model=WHISPER_MODEL,
                   device=DEVICE)
```

```python
res = whisper("./francisca.mp3")
display(IAudio("francisca.mp3"))

display(res["text"])
```
